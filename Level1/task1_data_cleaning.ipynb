# Level 1 – Task 1: Data Cleaning and Preprocessing
**Dataset:** Iris Dataset  
**Tools:** Python, pandas

## 1. Load the Dataset

```python
import pandas as pd

df = pd.read_csv('../Data Set For Task/1) iris.csv')
print('Shape:', df.shape)
df.head()
```

## 2. Introduce Realistic Dirty Data
The original Iris dataset is clean, so we simulate real-world issues: missing values, duplicates, and inconsistent formatting.

```python
import numpy as np

dirty_df = df.copy()

# Inject missing values
np.random.seed(42)
for col in ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']:
    missing_idx = np.random.choice(dirty_df.index, size=5, replace=False)
    dirty_df.loc[missing_idx, col] = np.nan

# Inject duplicate rows
dirty_df = pd.concat([dirty_df, dirty_df.iloc[[0, 1, 2]]], ignore_index=True)

# Inject inconsistent species formatting
dirty_df.loc[[10, 11, 12], 'species'] = ['Setosa', 'VERSICOLOR', 'virginica ']

print('Dirty dataset shape:', dirty_df.shape)
dirty_df.head(10)
```

## 3. Initial Inspection

```python
print('=== Data Types ===')
print(dirty_df.dtypes)

print('\n=== Missing Values ===')
print(dirty_df.isnull().sum())

print('\n=== Duplicate Rows ===')
print('Number of duplicates:', dirty_df.duplicated().sum())

print('\n=== Unique Species Values ===')
print(dirty_df['species'].unique())
```

## 4. Handle Missing Values
Numerical columns are imputed with the **median** (robust to outliers).

```python
clean_df = dirty_df.copy()

num_cols = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
for col in num_cols:
    median_val = clean_df[col].median()
    clean_df[col].fillna(median_val, inplace=True)

print('Missing values after imputation:')
print(clean_df.isnull().sum())
```

## 5. Remove Duplicate Rows

```python
clean_df.drop_duplicates(inplace=True)
clean_df.reset_index(drop=True, inplace=True)

print('Shape after removing duplicates:', clean_df.shape)
print('Duplicates remaining:', clean_df.duplicated().sum())
```

## 6. Standardize Inconsistent Categorical Values
Normalize the `species` column: lowercase and strip whitespace.

```python
clean_df['species'] = clean_df['species'].str.strip().str.lower()

print('Unique species after standardization:')
print(clean_df['species'].unique())
```

## 7. Final Cleaned Dataset Overview

```python
print('=== Final Shape:', clean_df.shape, '===')
print('\n=== Summary Statistics ===')
clean_df.describe()
```

```python
print('=== Species Distribution ===')
print(clean_df['species'].value_counts())
```

## 8. Save Cleaned Dataset

```python
clean_df.to_csv('iris_cleaned.csv', index=False)
print('Cleaned dataset saved as iris_cleaned.csv')
```

## Summary

| Step | Action | Result |
|------|--------|--------|
| Missing Values | Median imputation per numerical column | 0 missing values remaining |
| Duplicates | Removed duplicate rows | 3 duplicates removed |
| Inconsistent Format | Lowercased + stripped `species` column | 3 unique standardized values |
