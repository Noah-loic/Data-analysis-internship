# Level 1 – Task 3: Basic Data Visualization
**Dataset:** Iris Dataset (cleaned)  
**Tools:** Python, matplotlib, seaborn

## 1. Load the Dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

os.makedirs('plots', exist_ok=True)

df = pd.read_csv('iris_cleaned.csv')
df.head()
```

## 2. Bar Plot – Average Feature Values per Species

```python
species_means = df.groupby('species').mean(numeric_only=True)

species_means.plot(kind='bar', figsize=(10, 6), colormap='Set2', edgecolor='black')
plt.title('Average Feature Values per Species', fontsize=14)
plt.xlabel('Species', fontsize=12)
plt.ylabel('Average Value (cm)', fontsize=12)
plt.xticks(rotation=0)
plt.legend(title='Feature', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.savefig('plots/bar_plot.png', dpi=150)
plt.show()
```

## 3. Line Chart – Feature Values Across Samples

```python
num_cols = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']

plt.figure(figsize=(12, 5))
for col in num_cols:
    plt.plot(df.index, df[col], label=col.replace('_', ' ').title(), alpha=0.7)

plt.title('Feature Values Across All Samples', fontsize=14)
plt.xlabel('Sample Index', fontsize=12)
plt.ylabel('Value (cm)', fontsize=12)
plt.legend(title='Feature')
plt.tight_layout()
plt.savefig('plots/line_chart.png', dpi=150)
plt.show()
```

## 4. Scatter Plot – Petal Length vs Petal Width

```python
plt.figure(figsize=(8, 6))

colors = {'setosa': '#e74c3c', 'versicolor': '#2ecc71', 'virginica': '#3498db'}
for species, group in df.groupby('species'):
    plt.scatter(group['petal_length'], group['petal_width'],
                label=species.title(), color=colors[species], edgecolors='black', alpha=0.8)

plt.title('Petal Length vs Petal Width by Species', fontsize=14)
plt.xlabel('Petal Length (cm)', fontsize=12)
plt.ylabel('Petal Width (cm)', fontsize=12)
plt.legend(title='Species')
plt.tight_layout()
plt.savefig('plots/scatter_plot.png', dpi=150)
plt.show()
```

## 5. Summary

| Plot | Key Insight |
|------|-------------|
| Bar Plot | Virginica has the largest petal dimensions on average |
| Line Chart | Petal features show clear jumps between species groups |
| Scatter Plot | Setosa is fully separable by petal length and width |
